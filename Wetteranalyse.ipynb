# Wetteranalyse mit Python

Dieses Notebook untersucht Beispieldaten zu Temperatur, Niederschlag und Wind. Die Daten sind künstlich erstellt und stellen keine aktuellen Wetterdaten dar.

Benötigte Pakete: `pandas` und `matplotlib`. Falls sie fehlen, installiere sie mit `%pip install pandas matplotlib`.

## 1. Bibliotheken laden

Pandas verarbeitet die Wettertabelle; Matplotlib erstellt die Diagramme.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

## 2. Beispieldaten erstellen

Die Tabelle enthält Tageswerte für zehn aufeinanderfolgende Tage. Temperaturen sind in Grad Celsius, Niederschlag in Millimetern und Wind in km/h angegeben.

```python
wetter = pd.DataFrame({
    "datum": pd.date_range("2026-10-01", periods=10, freq="D"),
    "temperatur_min_c": [9, 11, 10, 8, 7, 9, 12, 13, 11, 8],
    "temperatur_max_c": [17, 19, 16, 14, 13, 18, 21, 22, 19, 15],
    "niederschlag_mm": [0, 1.2, 4.5, 0, 0, 2.1, 0, 0, 0.8, 6.3],
    "wind_kmh": [8, 12, 18, 10, 7, 9, 11, 14, 16, 20],
})

wetter["temperaturspanne_c"] = (
    wetter["temperatur_max_c"] - wetter["temperatur_min_c"]
)
display(wetter)
```

## 3. Kennzahlen berechnen

Mit `describe` erhältst du Anzahl, Mittelwert, Streuung sowie Minimum und Maximum der numerischen Spalten.

```python
kennzahlen = wetter.drop(columns="datum").describe().round(1)
display(kennzahlen)

print("Mittlere Höchsttemperatur:", round(wetter["temperatur_max_c"].mean(), 1), "°C")
print("Gesamtniederschlag:", round(wetter["niederschlag_mm"].sum(), 1), "mm")
```

## 4. Tage filtern

Hier werden warme Tage mit einer Höchsttemperatur ab 18 °C sowie Regentage mit mehr als 1 mm Niederschlag ausgewählt.

```python
warme_tage = wetter[wetter["temperatur_max_c"] >= 18]
regentage = wetter[wetter["niederschlag_mm"] > 1]

print("Warme Tage:")
display(warme_tage[["datum", "temperatur_max_c"]])

print("Regentage:")
display(regentage[["datum", "niederschlag_mm"]])
```

## 5. Wetter grafisch darstellen

Das Liniendiagramm vergleicht tägliche Tiefst- und Höchsttemperaturen. Das Balkendiagramm zeigt den Niederschlag.

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)

axes[0].plot(wetter["datum"], wetter["temperatur_min_c"], marker="o", label="Minimum")
axes[0].plot(wetter["datum"], wetter["temperatur_max_c"], marker="o", label="Maximum")
axes[0].set_ylabel("Temperatur (°C)")
axes[0].set_title("Temperaturverlauf")
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].bar(wetter["datum"], wetter["niederschlag_mm"], color="steelblue")
axes[1].set_ylabel("Niederschlag (mm)")
axes[1].set_title("Täglicher Niederschlag")

fig.autofmt_xdate()
fig.tight_layout()
plt.show()
```

## 6. Auffällige Tage zusammenfassen

Zum Abschluss werden der wärmste und der niederschlagsreichste Tag ermittelt.

```python
waermster_tag = wetter.loc[wetter["temperatur_max_c"].idxmax()]
nassester_tag = wetter.loc[wetter["niederschlag_mm"].idxmax()]

print(f"Wärmster Tag: {waermster_tag['datum']:%d.%m.%Y} mit {waermster_tag['temperatur_max_c']} °C")
print(f"Niederschlagsreichster Tag: {nassester_tag['datum']:%d.%m.%Y} mit {nassester_tag['niederschlag_mm']} mm")
```
