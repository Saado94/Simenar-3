# DWD-Wetterdaten Berlin-Dahlem, 2020 bis heute

Die CSV enthält tägliche Stationsmessungen des DWD für Berlin-Dahlem (Stations-ID `00403`). Der Datenstand reicht vom 01.01.2020 bis zum letzten verfügbaren DWD-Messtag. Nicht gemessene Werte sind leer und wurden nicht geschätzt.

Quelle: [DWD Climate Data Center](https://opendata.dwd.de/climate_environment/CDC/observations_germany/climate/daily/kl/).

## CSV laden

Die Datei enthält Datum, tägliche Mindest- und Mitteltemperatur, Niederschlagsmenge und Sonnenscheindauer. Temperaturen sind in °C, Niederschlag in mm und Sonnenschein in Stunden angegeben.

```python
from pathlib import Path

import pandas as pd

csv_datei = Path("DWD_Berlin_Dahlem_Tagesdaten_2020_heute.csv")
wetter = pd.read_csv(csv_datei, parse_dates=["datum"])

print(f"Zeitraum: {wetter['datum'].min():%d.%m.%Y} bis {wetter['datum'].max():%d.%m.%Y}")
print(f"Kalendertage: {len(wetter)}")
display(wetter.head())
```
