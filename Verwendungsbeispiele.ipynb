# Python-Bibliotheken: Verwendungsbeispiele

Dieses Notebook zeigt grundlegende Beispiele mit Python, NumPy, Matplotlib und Pandas. Die Beispiele verwenden kleine, direkt im Notebook definierte Datensätze.

Benötigte Pakete: `numpy`, `pandas` und `matplotlib`. Falls sie in deiner Umgebung fehlen, installiere sie vorab mit `%pip install numpy pandas matplotlib`.

## 1. Benötigte Bibliotheken importieren

Die drei Bibliotheken decken numerische Berechnungen, Tabellen und Diagramme ab.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## 2. Einfaches Python-Beispiel

Mit `sum` lässt sich die Summe aller Werte einer Liste direkt berechnen.

```python
werte = [12, 8, 15, 5]
summe = sum(werte)

print(f"Werte: {werte}")
print(f"Summe: {summe}")
```

## 3. Numerische Berechnungen mit NumPy

NumPy-Arrays erlauben kompakte elementweise Berechnungen. `mean` und `std` berechnen Mittelwert und Standardabweichung.

```python
messwerte = np.array([4.2, 5.1, 3.8, 6.0, 5.4])

print("Messwerte:", messwerte)
print("Plus 1:", messwerte + 1)
print("Mittelwert:", np.mean(messwerte))
print("Standardabweichung:", np.std(messwerte))
```

## 4. Visualisierung mit Matplotlib

Ein Liniendiagramm macht den Verlauf der Messwerte sichtbar. `xlabel`, `ylabel` und `title` beschriften die Grafik.

```python
tage = np.arange(1, len(messwerte) + 1)

plt.plot(tage, messwerte, marker="o", color="teal")
plt.xlabel("Messtag")
plt.ylabel("Messwert")
plt.title("Messwerte im Zeitverlauf")
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Datenanalyse mit Pandas

Ein `DataFrame` stellt tabellarische Daten dar. Hier berechnen wir Kennzahlen für die Messwerte und filtern anschließend die Zeilen mit einem Wert über 5.

```python
daten = pd.DataFrame({
    "Tag": tage,
    "Messwert": messwerte,
    "Ort": ["Nord", "Sued", "Nord", "West", "Sued"],
})

print("Datentabelle:")
display(daten)

print("Mittelwerte der numerischen Spalten:")
display(daten.mean(numeric_only=True))

print("Messwerte über 5:")
display(daten[daten["Messwert"] > 5])
```
